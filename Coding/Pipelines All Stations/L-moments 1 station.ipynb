```python
import os
import pandas as pd
import lmoments3 as lm

# Path to the CSV file
file_path = 'C:\\Users\\khfy3\\OneDrive\\Desktop\\Master Final Sem\\Coding\\Data\\Daily\\0600011RF.csv'

# Read data from the CSV file
data = pd.read_csv(file_path)['Value (mm)']

# Calculate the L-moments
l_moments = lm.lmom_ratios(data.dropna(), nmom=4)

# Print the results
print(f"File: {os.path.basename(file_path)}")
print(f"L1 (mean): {l_moments[0]}")
print(f"L2 (L-scale): {l_moments[1]}")
print(f"T3 (skewness ratio): {l_moments[2]}")
print(f"T4 (kurtosis ratio): {l_moments[3]}")
```
