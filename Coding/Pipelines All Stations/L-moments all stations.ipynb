```python
import os
import pandas as pd
import lmoments3 as lm

# Directory containing the CSV files
directory = 'C:\\Users\\khfy3\\OneDrive\\Desktop\\MSc Data Kemaman\\'

# Initialize a DataFrame to store results
results_df = pd.DataFrame(columns=['File', 'L1', 'L2', 'T3', 'T4'])

# Loop through each file in the directory
for filename in os.listdir(directory):
    if filename.endswith(".csv"):  # Check if the file is a CSV
        file_path = os.path.join(directory, filename)
        data = pd.read_csv(file_path)['Value (mm)']

        # Calculate the L-moments
        l_moments = lm.lmom_ratios(data.dropna(), nmom=4)

        # Append the results to the DataFrame
        results_df = results_df.append({
            'File': filename,
            'L1': l_moments['L1'],
            'L2': l_moments['L2'],
            'T3': l_moments['T3'],
            'T4': l_moments['T4']
        }, ignore_index=True)

# Save the DataFrame to an Excel file
excel_path = os.path.join(directory, 'L-moments_summary.xlsx')
results_df.to_excel(excel_path, index=False)

print("Results saved to:", excel_path)
```
